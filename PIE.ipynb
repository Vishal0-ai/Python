# Data Visualization By Pie Chart

```python
import matplotlib.pyplot as plt

labels = ["Sachin", "Sehwag", "Kohli", "Yuvraj"]
runs = [18000, 8000, 12000, 9500]
explode = [0.1, 0, 0, 0]
plt.pie(runs, labels=labels, explode=explode,shadow=True, # for percentage #)
plt.title("Exploded Pie Example")
plt.show()
```
