# Indexing and Slicing

```python
import numpy as np
arr = np.array([10, 20, 30, 40])
print(arr[0])  
print(arr[-1]) 
```

```python
arr = np.array([10, 20, 30, 40, 50])

print(arr[1:4])  
print(arr[:3])   
print(arr[::2])  
```
