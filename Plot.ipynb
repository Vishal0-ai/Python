```python
import matplotlib.pyplot as plt
```

```python
kohli = [1200, 1350, 1280, 1420, 1300, 1100, 1250, 1380, 1450, 1520]
rohit = [1050, 1180, 1250, 1320, 1400, 1180, 1300, 1240, 1350, 1420]
sehwag = [980, 1020, 1100, 1050, 1150, 950, 1080, 1120, 1200, 1180]
years = [2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024]

plt.plot(years, kohli, 'ro--', label="KOHLI",)
plt.plot(years, sehwag, 'g^:', label="SEHWAG")
plt.plot(years, rohit, label="ROHIT")
plt.xlabel("years")
plt.ylabel("runs")
plt.legend()
plt.style.use("seaborn-v0_8-darkgrid")
```
