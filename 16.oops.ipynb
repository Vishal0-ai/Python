## Objects

```python
class Student:
    name = "Vishal"
    age = 22


student1 = Student()

print(student1.name)
print(student1.age)
```

## __init__ Constructor


```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age


student1 = Student("Vishal", 22)

print(student1.name)
print(student1.age)
```

```python
# Encapsulation

class BankAccount:

    def __init__(self, balance):
        self.__balance = balance

    def get_balance(self):
        return self.__balance

    def deposit(self, amount):
        self.__balance += amount


account = BankAccount(1000)

account.deposit(500)

print(account.get_balance())
```

```python
# Inheritance

class Animal:

    def eat(self):
        print("Animal is eating")

class Dog(Animal):

    def bark(self):
        print("Dog is barking")


dog = Dog()

dog.eat()
dog.bark()
```

```python
# Method Overriding

class Animal:

    def sound(self):
        print("Animal makes a sound")


class Dog(Animal):

    def sound(self):
        print("Dog barks")


dog = Dog()

dog.sound()
```

```python
# Polymorphism
class Dog:

    def sound(self):
        print("Bark")


class Cat:

    def sound(self):
        print("Meow")


dog = Dog()
cat = Cat()

dog.sound()
cat.sound()
```

```python
# Abstraction

from abc import ABC, abstractmethod


class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass


class Car(Vehicle):

    def start(self):
        print("Car starts with a key")


car = Car()

car.start()
```
